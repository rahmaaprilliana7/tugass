Praktikum 01 – Sequential Search pada Data Angka

```python
def sequential_search(data, target):
  for i in range(len(data)):
      if data[i] == target:
        return f"Data {target} ditemukan di indeks {i}"
  return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(sequential_search(angka, 30))
print(sequential_search(angka, 100))
```

Praktikum 02 – Binary Search pada Data Angka

```python
def binary_search(data, target):
    left, right = 0, len(data)-1
    while left <= right:
      mid = (left + right) // 2
      if data[mid] == target:
          return f"Data {target} ditemukan di indeks {mid}"
      elif data[mid] < target:
          left = mid + 1
      else:
        right = mid - 1
    return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(binary_search(angka, 40))
print(binary_search(angka, 100))
```

Praktikum 03 – Sequential Search untuk Data Nama

```python
def sequential_search_nama(data, target):
  for i in range(len(data)):
      if data[i].lower() == target.lower():
          return f"Nama {target} ditemukan di indeks {i}"
  return f"Nama {target} tidak ditemukan"
nama = ["Andi", "Budi", "Cici", "Doni"]

print("Data:", nama)
print(sequential_search_nama(nama, "Cici"))
print(sequential_search_nama(nama, "Eka"))
```

Praktikum 04 – Studi Kasus Nilai Mahasiswa dengan Binary Search

Praktikum 06 – Kombinasi Sorting dan Searching

```python
import random
def insertion_sort(data):
    for i in range(1, len(data)):
         key = data[i]
         j = i - 1
         while j >= 0 and data[j] > key:
              data[j + 1] = data[j]
              j -= 1
         data[j + 1] = key
    return data

def binary_search(data, target):
    left, right = 0, len(data)-1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return f"{target} ditemukan di indeks {mid}"
        elif data[mid] < target:
             left = mid + 1
        else:
           right = mid - 1
    return f"{target} tidak ditemukan"

data = [random.randint(1, 100) for _ in range(15)]
print("Data acak sebelum diurutkan:", data)
urut = insertion_sort(data)
print("Data setelah diurutkan:", urut)
target = urut[7] # ambil angka ke-8
```

3. Variasi Sequential Search

```python
def sequential_search(data, target):
    comparisons = 0
    for i in range(len(data)):
        comparisons += 1
        if data[i] == target:
            return f"Data {target} ditemukan di indeks {i} setelah {comparisons} perbandingan"
    return f"Data {target} tidak ditemukan setelah {comparisons} perbandingan"


angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(sequential_search(angka, 30))
print(sequential_search(angka, 100))
```

4. Variasi Binary Search

```python
def binary_search(data, target):
    left, right = 0, len(data)-1
    steps = 0
    while left <= right:
        steps += 1
        mid = (left + right) // 2
        if data[mid] == target:
            return f"Data {target} ditemukan di indeks {mid} setelah {steps} langkah"
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return f"Data {target} tidak ditemukan setelah {steps} langkah"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(binary_search(angka, 40))
print(binary_search(angka, 100))
```

6. Studi Kasus Barang Toko

```python
def binary_search_barang(barang, target_harga):
    left, right = 0, len(barang) - 1
    steps = 0
    while left <= right:
        steps += 1
        mid = (left + right) // 2
        if barang[mid][1] == target_harga:
            return f"Barang '{barang[mid][0]}' dengan harga {target_harga} ditemukan di indeks {mid} setelah {steps} langkah"
        elif barang[mid][1] < target_harga:
            left = mid + 1
        else:
            right = mid - 1
    return f"Barang dengan harga {target_harga} tidak ditemukan setelah {steps} langkah"

def main():
    n = int(input("Masukkan jumlah barang: "))
    barang = []
    for i in range(n):
        nama = input(f"Nama barang ke-{i+1}: ")
        harga = int(input(f"Harga barang ke-{i+1}: "))
        barang.append((nama, harga))

    barang.sort(key=lambda x: x[1])
    print("\nDaftar barang setelah diurutkan berdasarkan harga:")
    for nama, harga in barang:
        print(f"{nama} - Rp{harga}")

    target = int(input("\nMasukkan harga barang yang ingin dicari: "))
    hasil = binary_search_barang(barang, target)
    print(hasil)


main()
```

7. Analisis Kompleksitas

```python
import time
import random

def sequential_search(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1

def binary_search(data, target):
    left, right = 0, len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def main():
    sizes = [100, 1000, 10000]
    print(f"{'Ukuran':<10}{'Sequential (detik)':<25}{'Binary (detik)':<25}")
    print("-" * 60)

    for size in sizes:
        data = list(range(size))
        target = random.choice(data)

        start = time.time()
        sequential_search(data, target)
        seq_time = time.time() - start

        start = time.time()
        binary_search(data, target)
        bin_time = time.time() - start

        print(f"{size:<10}{seq_time:<25.10f}{bin_time:<25.10f}")

main()
```

8. Proyek Mini

```python
def binary_search_mahasiswa(mahasiswa, target_nilai):
    left, right = 0, len(mahasiswa) - 1
    steps = 0
    while left <= right:
        steps += 1
        mid = (left + right) // 2
        if mahasiswa[mid][1] == target_nilai:
            return f"Mahasiswa '{mahasiswa[mid][0]}' dengan nilai {target_nilai} ditemukan di indeks {mid} setelah {steps} langkah"
        elif mahasiswa[mid][1] < target_nilai:
            left = mid + 1
        else:
            right = mid - 1
    return f"Nilai {target_nilai} tidak ditemukan setelah {steps} langkah"

def main():
    n = int(input("Masukkan jumlah mahasiswa: "))
    mahasiswa = []
    for i in range(n):
        nama = input(f"Nama mahasiswa ke-{i+1}: ")
        nilai = int(input(f"Nilai mahasiswa ke-{i+1}: "))
        mahasiswa.append((nama, nilai))

    mahasiswa.sort(key=lambda x: x[1])
    print("\nDaftar mahasiswa setelah diurutkan berdasarkan nilai:")
    for nama, nilai in mahasiswa:
        print(f"{nama} - {nilai}")

    target = int(input("\nMasukkan nilai yang ingin dicari: "))
    hasil = binary_search_mahasiswa(mahasiswa, target)
    print(hasil)

main()
```
