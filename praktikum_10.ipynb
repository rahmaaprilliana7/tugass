praktikum 01 - bubble sort

```python
def bubble_sort(data):
    n = len(data)
    for i in range (n):
        for j in range (0, n - i - 1):
            if data [j] > data [j + 1]:
                data [j] , data [j + 1 ] = data[j + 1], data [j]
    return data

angka = [64, 34, 25, 12, 22, 11, 90]
print ("data sebelum diurutkan:", angka)
print ("data setelah diurutkan:", bubble_sort (angka))
```

praktikum 2 - selection sort

```python
def selection_sort(data):
    n = len (data)
    for i in range (n):
        min_index = i
        for j in range (i + 1, n):
            if data[j] < data[min_index]:
                min_index = j
            data [i], data [min_index] = data [min_index], data[i]
    return data

angka = [64, 34, 25, 12, 22]
print ("data sebelum diurutkan:", angka)
print ("data setelah diurutkan:", selection_sort (angka))
```

praktikum 03 - insertion sort

```python
def insertion_sort(data):
    n = len (data)
    for i in range (1, n):
        key = data[i]
        j = i-1
        while j >= 0 and data[j] > key:
            data [j+1] = data[j]
            j -= 1
        data[j+1] = key
    return data

angka = [12, 11, 13, 5, 6]
print ("data sebelum diurutkan:", angka)
print ("data setelah diurutkan:", insertion_sort (angka))
```

7. Analisis Kompleksitas

```python
import random

def bubble_sort(arr):
    comparisons = 0
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return comparisons


def selection_sort(arr):
    comparisons = 0
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return comparisons


def insertion_sort(arr):
    comparisons = 0
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                j -= 1
            else:
                break
        arr[j + 1] = key
    return comparisons


def test(size):
    random.seed(1)  # supaya hasil konsisten
    data = random.sample(range(size * 10), size)

    print(f"Ukuran {size}:")
    print(f"  Bubble Sort    -> {bubble_sort(data.copy())} perbandingan")
    print(f"  Selection Sort -> {selection_sort(data.copy())} perbandingan")
    print(f"  Insertion Sort -> {insertion_sort(data.copy())} perbandingan")
    print()


for n in [10, 50, 100]:
    test(n)
```

8. Proyek Mini

```python
def input_barang():
    barang = []
    n = int(input("Masukkan jumlah barang: "))
    for i in range(n):
        nama = input(f"Nama barang ke-{i+1}: ")
        harga = int(input(f"Harga barang ke-{i+1}: "))
        barang.append((nama, harga))
    return barang

def tampilkan(judul, barang):
    print(f"\n{judul}:")
    for nama, harga in barang:
        print(f"{nama} - Rp{harga}")

def main():
    barang = input_barang()1

    tampilkan("Data Asli", barang)

    nama_asc = sorted(barang, key=lambda x: x[0].lower())
    tampilkan("Urut Nama Ascending", nama_asc)

    nama_desc = sorted(barang, key=lambda x: x[0].lower(), reverse=True)
    tampilkan("Urut Nama Descending", nama_desc)

    harga_asc = sorted(barang, key=lambda x: x[1])
    tampilkan("Urut Harga Ascending", harga_asc)

    harga_desc = sorted(barang, key=lambda x: x[1], reverse=True)
    tampilkan("Urut Harga Descending", harga_desc)

if __name__ == "__main__":
    main()
```
