```python
%%writefile /content/drive/MyDrive/tugas/laporan.py

def tampilkan_ringkasan(skor, jumlah_soal):
    print("=== Ringkasan Hasil Kuis ===")
    print(f"Jumlah Soal   : {jumlah_soal}")
    print(f"Jawaban Benar : {skor}")
    print(f"Jawaban Salah : {jumlah_soal - skor}")
    print(f"Skor Akhir    : {skor}/{jumlah_soal}")
```

```python

import random
import sys
sys.path.append('/content/drive/MyDrive/tugas/')

import laporan

def kuis():
  skor = 0
  jumlah_soal = 5

  for i in range(0,5):
    a= random.randint(1,10)
    b= random.randint(1,10)
    print(f"(soal {i+1}) {a} + {b} = ...")
    jawaban = int(input("masukkan jawaban anda: "))

    if jawaban == a+b:
      print("jawaban anda benar")
      skor  += 1

    else :
      print("jawaban anda salah")

  laporan.tampilkan_ringkasan(skor, jumlah_soal)

if __name__ == "__main__":
  kuis()
```
